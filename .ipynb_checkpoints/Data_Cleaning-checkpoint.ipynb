# Data cleaning

In this document I will clean the data and justify my actions as I progress

To start we will read the csv and create a backup dataframe with the origional data so that we can compare it if needed

```python
import numpy as np
import pandas as pd
import re
```

```python
data = pd.read_csv("LaptopData.csv")
data_backup = data.copy()
data.head()
```

The first thing we can remove the unnamed column as this is clearly an index that is no longer nessecary

```python
data = data.drop(data.columns[0],axis=1)
data.head()
```

Through exploration of the data I have found what seams to be seperator rows of blank data

```python
data.isna().sum()
```

```python
data.iloc[20]
```

```python
data = data.dropna()
```

```python
data.isna().sum()
```

```python
data.iloc[20]
```

I've now removed the blank rows as they held no data

Now I need to standardise the "ScreenResolution" column

```python
data["ScreenResolution"].size
```

```python
type(data["ScreenResolution"].iloc[0])
```

```python
res = data["ScreenResolution"].tolist()
restest = data.copy()
print(type(res))
res[0]
```

```python
templist = []
for item in res:
    templist.append(re.sub(r'[^0-9x]', '', item))
templist[0]
```

I now have a list of the resolutions standardised 

```python
if len(templist) != len(restest):
    print("length missmatch")
else:
    restest["ScreenResolution"] = templist
    print("complete")
```

```python
restest.head()
```

```python
data = restest.copy()
data.head()
```

I now need to do a similar process with memory

```python
mem = data["Memory"].tolist()
memtest = data.copy()
print(type(mem))
mem[0]
templist = []
for item in mem:
    templist.append(re.sub(r'[^0-9GB]', '', item))
templist[0]
```

```python
if len(templist) != len(memtest):
    print("length missmatch")
else:
    memtest["Memory"] = templist
    print("complete")
```

```python
memtest.head()
```

After some data exploration I have found that the Memory column is very messy (multiple SSDs and some in TB compared to GB) and as it's not overall relevent to the current analysis I will be dropping this column

```python
data = data.drop("Memory", axis=1)
data.head()
```

Now all that remains is to write it back to a new csv to be accessed from other notebooks

```python
data.to_csv("data.csv")
"completed"
```
